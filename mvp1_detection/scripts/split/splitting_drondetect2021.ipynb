# Splitting

```python
import os
import shutil
import random
from pathlib import Path

# Original paths
from pathlib import Path

dataset_path = Path("/kaggle/input/dronedetection2021-yolo-unified/DroneDetection2021_YOLO_Unified/Drones")

images_dir = Path("/kaggle/input/datasets/reemmousa87/dronedetection2021/DroneDetection2021_YOLO_Unified/Drones/images/all")
labels_dir = Path("/kaggle/input/datasets/reemmousa87/dronedetection2021/DroneDetection2021_YOLO_Unified/Drones/labels/all")


# New folders
output_path = Path("/kaggle/working/Drones")

for split in ["train", "val", "test"]:
    (output_path / "images" / split).mkdir(parents=True, exist_ok=True)
    (output_path / "labels" / split).mkdir(parents=True, exist_ok=True)


# Get images
images = list(images_dir.glob("*.jpg"))

print("Total images:", len(images))


# Shuffle
random.seed(42)
random.shuffle(images)


# Split ratio
train_ratio = 0.7
val_ratio = 0.2
test_ratio = 0.1


train_end = int(len(images) * train_ratio)
val_end = int(len(images) * (train_ratio + val_ratio))


train_images = images[:train_end]
val_images = images[train_end:val_end]
test_images = images[val_end:]


splits = {
    "train": train_images,
    "val": val_images,
    "test": test_images
}


# Copy images + labels together
for split, files in splits.items():

    for img_path in files:

        # Copy image
        shutil.copy(
            img_path,
            output_path / "images" / split / img_path.name
        )

        # Find matching label
        label_name = img_path.stem + ".txt"
        label_path = labels_dir / label_name

        if label_path.exists():
            shutil.copy(
                label_path,
                output_path / "labels" / split / label_name
            )
        else:
            print("Missing label:", label_name)


print("Dataset split completed!")
```

```python
for folder in ["train", "val", "test"]:
    print(
        folder,
        "images:",
        len(list((output_path/"images"/folder).glob("*.jpg"))),
        "labels:",
        len(list((output_path/"labels"/folder).glob("*.txt")))
    )
```

```python
yaml_content = """
path: /kaggle/working/Drones

train: images/train
val: images/val
test: images/test

names:
  0: drone
"""

with open("/kaggle/working/Drones/data.yaml", "w") as f:
    f.write(yaml_content)
```
